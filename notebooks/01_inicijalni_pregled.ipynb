U ovoj svesci prvi put zavirujemo u sirove podatke. Ideja je da se upoznamo sa strukturom skupa, pogledamo tipove podataka i uočimo eventualne probleme (nedostajuće vrednosti, duplikate, besmislene vrednosti) pre nego što bilo šta diramo.

Bitna napomena: ovde podatke samo posmatramo i dijagnostikujemo, ne menjamo ih. Čišćenje i obrada dolaze u kasnijim sveskama, tek nakon podele na trening i test skup.

## 1. Učitavanje podataka i biblioteka

Koristimo pandas, standardni alat za rad sa tabelarnim podacima u Python-u. Skup učitavamo iz CSV fajla u DataFrame — u suštini tabelu sa redovima i kolonama.

```python
#Prvi put importujemo biblioteku numpy (np) - za numeričke operacije; koristićemo ga kasnije za rad sa NaN vrednostima.
import pandas as pd
import numpy as np

# Podešavanje da pandas prikazuje SVE kolone kada ispisujemo tabelu.
# Bez ovoga, pandas skraćuje prikaz i sakriva srednje kolone sa "...".
pd.set_option("display.max_columns", None)


df = pd.read_csv("../WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Podaci su uspešno učitani.")
```

## 4. Informacije o kolonama i tipovima podataka

`info()` nam daje pregled svih kolona: naziv, broj popunjenih (ne-null) vrednosti i tip podatka (`dtype`).

Ovo je zgodno za rano hvatanje problema. Najviše nas zanima da li se tipovi poklapaju sa onim što očekujemo — recimo, `TotalCharges` bi trebalo da bude broj, pa ako je učitana kao tekst (`object`), to je već znak da nešto nije kako treba.

```python
# df.info() ispisuje pregled: naziv kolone, broj ne-null vrednosti i tip podataka.
df.info()
```

## 5. Provera nedostajućih vrednosti



```python
# df.isnull() vraća tabelu True/False (True = vrednost nedostaje).
# .sum() sabira po kolonama: koliko True (nedostajućih) ima u svakoj koloni.
print("Broj nedostajućih vrednosti po kolonama:")
print(df.isnull().sum())
```

## 6. Skrivene nedostajuće vrednosti

Prethodna provera je pokazala da nema nedostajućih vrednosti, ali to zna da prevari. U praksi se nedostajuće vrednosti nekad kriju kao prazni razmaci (`" "`) umesto pravih `NaN` oznaka. `isnull()` ih ne uhvati jer su tehnički gledano tekst, a ne prazne vrednosti.

Zato dodatno proveravamo da li neka kolona sadrži vrednosti koje su samo prazan razmak.

```python
# Poredimo svaku ćeliju sa praznim razmakom " ".
# (df == " ") vraća tabelu True/False (True = ćelija sadrži tačno jedan razmak).
# .sum() sabira po kolonama koliko takvih "praznih" vrednosti ima.
print("Broj vrednosti koje su prazan razmak, po kolonama:")
print((df == " ").sum())
```

## 7. Analiza problema sa kolonom `TotalCharges`

Sad je jasno zašto je `TotalCharges` prepoznata kao tekst (`object`) umesto broja: pošto 11 vrednosti sadrži prazan razmak (a to nije broj), pandas je celu kolonu protumačio kao tekstualnu.


```python
# pd.to_numeric() pokušava da pretvori kolonu u broj.
# errors="coerce" znači: vrednosti koje se NE mogu pretvoriti u broj postaju NaN umesto da izazovu grešku.
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Ponovo proveravamo nedostajuće vrednosti — sada bi TotalCharges trebalo da pokaže 11 NaN vrednosti.
print("Broj nedostajućih vrednosti nakon konverzije TotalCharges:")
print(df.isnull().sum())
```

Zaključak je da `TotalCharges` treba da bude numerička kolona, pa je konvertujemo.

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")
```

## 8. Analiza korisnika sa nedostajućim `TotalCharges`

Sad kad je `TotalCharges` numerička, možemo da vidimo tačno koliko korisnika ima nedostajuću vrednost. Hoćemo i da razumemo zašto baš oni — da li je posredi slučajna greška ili tu postoji neki obrazac.

```python
print("Nedostajuće vrednosti nakon konverzije TotalCharges:")
print(df.isnull().sum())
print()
```

### Ključna opservacija

Svih 11 korisnika sa nedostajućom vrednošću `TotalCharges` ima `tenure = 0`. To su, dakle, korisnici koji su tek počeli da koriste uslugu, pa u trenutku formiranja skupa za njih još nije bilo evidentirane ukupne naplate.

Ovo nam dosta govori. Nije u pitanju slučajna greška u merenju niti namerno preskočeno polje — vrednost jednostavno logički ne postoji za te korisnike (nova pretplata, naplate još nije bilo).

Šta ćemo sa ovim redovima, odlučićemo kasnije, nakon podele na trening i test skup, da ne bismo napravili curenje informacija (data leakage).

Zaključak je da nedostajuće vrednosti ipak postoje u skupu, pa u sledećem koraku utvrđujemo koje su tačno.

```python
print("Korisnici sa nedostajućim TotalCharges:")
print(df[df["TotalCharges"].isnull()])
print()
```

## 11. Provera kategorijskih promenljivih

### 11.1 SeniorCitizen — binarna kategorija u brojnom obliku

`SeniorCitizen` je zapisana kao broj (0 ili 1), ali je suštinski binarna kategorija: da li je korisnik stariji građanin ili ne. Zato umesto statistike gledamo raspodelu po vrednostima, da vidimo koliko korisnika pada u svaku grupu.

```python
print("Raspodela SeniorCitizen:")
print(df["SeniorCitizen"].value_counts())
print()
```

Većina korisnika (~5.900, oko 84%) nisu stariji građani, dok ih je manjina (~1.140, oko 16%) u toj grupi. Ova nesrazmera će nam biti bitna kasnije — proverićemo da li starost korisnika ima veze sa odlivom.

Izdvajamo sve kolone koje imaju tekstualne vrednosti.

Za svaku kategoriju prikazujemo sve *`postojece`* kategorije i *`broj`* korisnika u svakoj kategoriji

### 11.2 Sve kategorijske promenljive

Prolazimo kroz sve kolone tekstualnog tipa (`object`) i za svaku ispisujemo postojeće kategorije zajedno sa brojem korisnika. Tako vidimo tačno koje se vrednosti javljaju, da li ima nekih neočekivanih ili nekonzistentnih kategorija i koliko su kategorije izbalansirane (da li negde jedna vrednost dominira).

`customerID` preskačemo — tih 7043 jedinstvenih identifikatora nema šta da nam kaže u ovakvom prikazu.

```python
# Izdvajamo sve kolone tekstualnog tipa.
categorical_columns = df.select_dtypes(include="object").columns

# Za svaku kategorijsku kolonu prikazujemo raspodelu.
# Izuzimamo customerID jer predstavlja jedinstveni identifikator korisnika,
# a ne kategoriju za analizu.
for column in categorical_columns:
    if column == "customerID":
        continue
    print(f"\n{column}:")
    print(df[column].value_counts())
```

**Kolone sa tri vrednosti (`Yes` / `No` / `"No internet service"` i sl.**
`MultipleLines`, `OnlineSecurity`, `OnlineBackup`, `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies`.

`"No internet service"` i `"No"` nisu isto:

`"No internet service"` — korisnik uopšte nema internet
Ova razlika će biti bitna za feature engineering, pa ćemo o tome (spajati ih ili ne) odlučiti kasnije.

**Kolone sa više kategorija:**

## 12. Zaključak

Kroz inicijalni pregled skupa `Telco Customer Churn` došli smo do sledećeg.

**Struktura skupa:** 7.043 korisnika i 21 kolona, koje smo podelili u 4 grupe — demografske, usluge, podaci o nalogu i plaćanju, i ciljna promenljiva.

**Problemi koje ćemo rešavati:**
1. Skrivene nedostajuće vrednosti u `TotalCharges` — 11 korisnika (0.16%) sa `tenure = 0`. Posle konverzije u broj ti redovi imaju `NaN`. Šta ćemo sa njima, odlučujemo tek nakon podele na trening/test.
2. `SeniorCitizen` je zapisana kao 0/1 iako je binarna kategorija. Razmislićemo o konverziji u `No`/`Yes` radi konzistentnosti.
3. Vrednosti `"No internet service"` i `"No phone service"` nisu isto što i `"No"`. O njihovom tretmanu odlučujemo u fazi pripreme podataka.
4. Nebalansirane klase u `Churn` (~73% No, ~27% Yes), što dosta utiče na izbor metrika i na pristup modeliranju.

**Ono što je u redu:**
- Nema duplikata (ni potpunih redova ni po `customerID`)
- Numeričke kolone nemaju nelogične vrednosti (negativne ili van opsega)
- Kategorijske kolone imaju konzistentne vrednosti, bez tipfelera
