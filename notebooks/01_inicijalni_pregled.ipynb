## 1. Učitavanje podataka i biblioteka

Za rad koristimo biblioteku **pandas**, standardni alat za rad sa tabelarnim
podacima u Python-u. Učitavamo skup podataka iz CSV fajla u strukturu koja se
zove **DataFrame** (tabela sa redovima i kolonama).

```python
# Uvozimo biblioteke potrebne za ovu fazu.
# pandas (pd) - za rad sa tabelama.
# numpy (np) - za numeričke operacije; koristićemo ga kasnije za rad sa NaN vrednostima.
import pandas as pd
import numpy as np

# Podešavanje da pandas prikazuje SVE kolone kada ispisujemo tabelu.
# Bez ovoga, pandas skraćuje prikaz i sakriva srednje kolone sa "...".
pd.set_option("display.max_columns", None)

# Učitavamo CSV fajl u DataFrame df.
# "../" znači da izlazimo iz foldera notebooks/ u glavni folder repoa.
df = pd.read_csv("../WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Podaci su uspešno učitani.")
```

## 5. Provera nedostajućih vrednosti

Standardnu proveru nedostajućih vrednosti radimo funkcijom `isnull()`, koja za
svaku ćeliju vraća `True` ako je vrednost nedostajuća (`NaN`), a `False` inače.
Sabiranjem tih vrednosti po kolonama dobijamo broj nedostajućih vrednosti u svakoj koloni.

```python
# df.isnull() vraća tabelu True/False (True = vrednost nedostaje).
# .sum() sabira po kolonama: koliko True (nedostajućih) ima u svakoj koloni.
print("Broj nedostajućih vrednosti po kolonama:")
print(df.isnull().sum())
```

## 6. Skrivene nedostajuće vrednosti

Prethodna provera pokazala je da nema nedostajućih vrednosti — ali to može biti
obmanjujuće. U realnim podacima nedostajuće vrednosti se ponekad kriju kao
**prazni razmaci** (`" "`) umesto pravih `NaN` oznaka. Funkcija `isnull()` ih
ne prepoznaje jer su tehnički tekst, a ne prazne vrednosti.

Zato dodatno proveravamo da li neka kolona sadrži vrednosti koje su samo prazan razmak.

```python
# Poredimo svaku ćeliju sa praznim razmakom " ".
# (df == " ") vraća tabelu True/False (True = ćelija sadrži tačno jedan razmak).
# .sum() sabira po kolonama koliko takvih "praznih" vrednosti ima.
print("Broj vrednosti koje su prazan razmak, po kolonama:")
print((df == " ").sum())
```

## 7. Analiza problema sa kolonom `TotalCharges`

Sada razumemo zašto je `TotalCharges` prepoznata kao tekst (`object`) umesto broja:
pošto 11 vrednosti sadrži prazan razmak (koji nije broj), pandas je celu kolonu
protumačio kao tekstualnu.

Da bismo mogli da analiziramo ovu kolonu numerički, moramo je konvertovati u
broj. Vrednosti koje se ne mogu konvertovati (prazni razmaci) pretvorićemo u
prave `NaN` oznake, koje potom možemo pravilno da analiziramo.

```python
# pd.to_numeric() pokušava da pretvori kolonu u broj.
# errors="coerce" znači: vrednosti koje se NE mogu pretvoriti u broj
# (prazni razmaci) postaju NaN umesto da izazovu grešku.
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Ponovo proveravamo nedostajuće vrednosti — sada bi TotalCharges trebalo da
# pokaže 11 NaN vrednosti (bivši prazni razmaci).
print("Broj nedostajućih vrednosti nakon konverzije TotalCharges:")
print(df.isnull().sum())
```

**Ovde zakljucujemo da TotalCharges bi trebalo da bude `numericka vrednost` sto nam govori da je treba da je konvertujemo**. To mozemo uraditi preko ugradjene funkcije u biblioteci `pandas`. Vrednosti koje ne mogu da se konvertuju postaju `NaN` vrednosti.

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")
```

## 8. Provera nedostajucih vrednosti nakon `konverzije` tipa

```python
print("Nedostajuće vrednosti nakon konverzije TotalCharges:")
print(df.isnull().sum())
print()
```

Ovde dolazimo do zakljucka da je `postoje` nedostajucih vrednosti u nasem skupu podataka i u sledecem koraku da utvrdimo koje su to vrednosti

```python
print("Korisnici sa nedostajućim TotalCharges:")
print(df[df["TotalCharges"].isnull()])
print()
```

## 9. Brisanje observacija
